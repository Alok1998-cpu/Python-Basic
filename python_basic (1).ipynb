Python Basic


1- What is Python, and why is it popular ?
   - Python is a versatile, high level, interpreted programming language known for its readability and ease of use. Its popularity stems from several factors, including a large and active community, extensive libraries, and its use in various fields like web development, data science, and machine learning.                
2- What is an interpreter in Python ?
   - In Python, an interpreter is a program that reads and executes Python code line by line. It translates the high-level code into machine-readable bytecode, which is then executed. This process allows Python to run without requiring a pre-compilation step like compiled languages.
3-   What are pre-defined keywords in Python ?
     - Keywords in Python are reserved words that have predefined meanings and purposes within the language. These keywords cannot be used as identifiers (variable names, function names, etc.) because they are integral to Python's syntax and structure.
There are 36 keywords in Python. They are categorized based on their purpose
4- Can keywords be used as variable names ?
   - No, keywords cannot be used as variable names in most programming languages.
5-  What is mutability in Python ?
    - Mutability in Python refers to whether the state of an object can be modified after it is created. Mutable objects can be changed, while immutable objects cannot.
6-  Why are lists mutable, but tuples are immutable ?
    - Tuples and lists are the same in every way except two: tuples use parentheses instead of square brackets, and the items in tuples cannot be modified (but the items in lists can be modified). We often call lists mutable (meaning they can be changed) and tuples immutable (meaning they cannot be changed).
7-  What is the difference between “==” and “is” operators in Python ?
    - The == and is operators in Python serve different purposes when comparing objects.
== (Equality Operator):
Checks if the values of two objects are equal.
Compares the content or data held by the objects.
Memory locations of the objects are irrelevant.
Two distinct objects with the same content will return True with the == operator.
is (Identity Operator):
Checks if two variables refer to the same object in memory.
Compares the memory addresses of the objects.
Two variables point to the exact same object in memory when is returns True.
Often used to check if a variable is None.
8- What are logical operators in Python ?
  - Logical operators in Python are used to combine or modify conditional statements. They operate on boolean values (True or False) and return a boolean result. There are three main logical operators in Python:
and: Returns True if both operands are True; otherwise, it returns False.
9-  What is type casting in Python ?
    - type casting, also known as type conversion, is the process of changing the data type of a variable to another data type. Python supports two types of type casting: implicit and explicit.
10- What is the difference between implicit and explicit type casting ?
    - Implicit type casting, also known as widening or upcasting, happens automatically when converting a smaller data type to a larger one, such as from an int to a double. Explicit type casting, also known as narrowing or downcasting, requires the programmer to manually specify the conversion, for example, converting a double to an int, which may result in data loss.
11-  What is the purpose of conditional statements in Python ?
    - Conditional statements in Python are used to control the flow of a program based on specific conditions. They allow different blocks of code to be executed depending on whether a condition is true or false. This provides a way to make decisions in the program and execute different code based on those decisions.
12- How does the elif statement work ?
    - 'Elif' stands for 'else if' and is used in Python programming to test multiple conditions. It is written following an if statement in Python to check an alternative condition if the first condition is false. The code block under the elif statement will be executed only if its condition is true.
13- What is the difference between for and while loops ?
    - The primary difference between for and while loops lies in their control flow: for loops are used when the number of iterations is known in advance, and while loops are used when the number of iterations depends on a condition that may not be known beforehand.
14- Describe a scenario where a while loop is more suitable than a for loop ?
    - In programming, for loops are best used when you know the number of iterations ahead of time, whereas a while loop is best used when you don't know the number of iterations in advance. Both methods can help you iterate through your code.

```python
# write a program that checks if given word is python keyword
import keyword

def is_keyword(word):
    """Checks if a given word is a Python keyword.

    Args:
        word: The word to check.

    Returns:
        True if the word is a keyword, False otherwise.
    """
    return keyword.iskeyword(word)

# Example usage
word1 = "for"
word2 = "hello"

print(f"'{word1}' is a keyword: {is_keyword(word1)}")
print(f"'{word2}' is a keyword: {is_keyword(word2)}")
```

```python
# create a list and tuple in python,and demonstrate how attempting to change an element works diffrently for each
# Creating a list
my_list = [1, 2, 3]
print("Original list:", my_list)

# Modifying an element in the list
my_list[1] = 4
print("Modified list:", my_list)

# Creating a tuple
my_tuple = (1, 2, 3)
print("Original tuple:", my_tuple)

# Attempting to modify an element in the tuple (this will raise an error)
try:
    my_tuple[1] = 4
except TypeError as e:
    print("Error:", e)
```

```python
# Write a function to demonstrate the behavior of mutable and immutable arguments
def modify_arguments(immutable_arg, mutable_arg):
    """
    Demonstrates the behavior of mutable and immutable arguments.

    Args:
      immutable_arg: An immutable object (e.g., integer, string, tuple).
      mutable_arg: A mutable object (e.g., list, dictionary).
    """
    # Attempt to modify the immutable argument
    try:
        immutable_arg = immutable_arg + 1
    except TypeError:
        print("Cannot modify immutable arguments directly")

    # Modify the mutable argument
    mutable_arg.append(4)
    mutable_arg[0] = 99

    print("Inside function:")
    print("Immutable:", immutable_arg)
    print("Mutable:", mutable_arg)


# Example usage
immutable_value = 5
mutable_value = [1, 2, 3]

print("Before function call:")
print("Immutable:", immutable_value)
print("Mutable:", mutable_value)

modify_arguments(immutable_value, mutable_value)

print("After function call:")
print("Immutable:", immutable_value)
print("Mutable:", mutable_value)
```

```python
#  Write a program that performs basic arithmetic operations on two user-input numbers
def add(x, y):
  """
  Adds two numbers.
  """
  return x + y

def subtract(x, y):
  """
  Subtracts two numbers.
  """
  return x - y

def multiply(x, y):
  """
  Multiplies two numbers.
  """
  return x * y

def divide(x, y):
  """
  Divides two numbers.
  """
  if y == 0:
    return "Error: Cannot divide by zero."
  else:
    return x / y

# Get input from the user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")

# Perform the operation
if operation == "+":
  result = add(num1, num2)
elif operation == "-":
  result = subtract(num1, num2)
elif operation == "*":
  result = multiply(num1, num2)
elif operation == "/":
  result = divide(num1, num2)
else:
  result = "Invalid operation"

# Print the result
print("Result:", result)
```

```python
#  Write a program to demonstrate the use of logical operators
#include <stdio.h>
#include <stdbool.h>

int main() {
    // Variables
    int age = 25;
    bool hasLicense = true;
    bool isRegistered = false;

    // Logical operators demonstration
    // AND operator (&&) - returns true if both operands are true
    if (age >= 16 && hasLicense) {
        printf("You can drive.\n");
    } else {
        printf("You cannot drive.\n");
    }

    // OR operator (||) - returns true if at least one operand is true
    if (isRegistered || hasLicense) {
        printf("You are either registered or have a license.\n");
    } else {
        printf("You are neither registered nor have a license.\n");
    }

    // NOT operator (!) - returns the opposite value
    if (!isRegistered) {
        printf("You are not registered.\n");
    } else {
        printf("You are registered.\n");
    }

    return 0;
}
```

```python
#  Write a Python program to convert user input from string to integer, float, and boolean types
def convert_input():
    user_input = input("Enter a value: ")

    try:
        # Convert to integer
        integer_value = int(user_input)
        print(f"Integer: {integer_value}, Type: {type(integer_value)}")
    except ValueError:
        print("Cannot convert to integer.")

    try:
        # Convert to float
        float_value = float(user_input)
        print(f"Float: {float_value}, Type: {type(float_value)}")
    except ValueError:
        print("Cannot convert to float.")

    # Convert to boolean
    if user_input.lower() == "true":
        boolean_value = True
    elif user_input.lower() == "false":
        boolean_value = False
    else:
        boolean_value = bool(user_input) # Non-empty strings evaluate to True, empty strings evaluate to False
    print(f"Boolean: {boolean_value}, Type: {type(boolean_value)}")

convert_input()
```

```python
# Write a program that checks if a number is positive, negative, or zero
def check_number(num):
  """
  Checks if a number is positive, negative, or zero.

  Args:
    num: The number to check.

  Returns:
    A string indicating whether the number is positive, negative, or zero.
  """
  if num > 0:
    return "Positive"
  elif num < 0:
    return "Negative"
  else:
    return "Zero"

# Example usage
number = 5
result = check_number(number)
print(f"The number {number} is {result}.")  # Output: The number 5 is Positive.

number = -3
result = check_number(number)
print(f"The number {number} is {result}.")  # Output: The number -3 is Negative.

number = 0
result = check_number(number)
print(f"The number {number} is {result}.")  # Output: The number 0 is Zero.
```

```python
# Write a Python program to calculate the factorial of a number provided by the user using a while loop

num = int(input("Enter a non-negative integer: "))

if num < 0:
    print("Factorial is not defined for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    factorial = 1
    i = 1
    while i <= num:
        factorial *= i
        i += 1
    print("The factorial of", num, "is", factorial)
```
